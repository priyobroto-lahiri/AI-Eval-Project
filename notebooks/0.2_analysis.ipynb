```python
import pandas as pd
df = pd.read_csv("../results/my_output.csv")
print(f"Total Rows Loaded: {len(df)}")
df.head() # This displays the first 5 rows in a nice table
```

```python
avg_latency = df['latency'].mean()
total_tokens = df['prompt_tokens'].sum() + df['response_tokens'].sum()
print(f"📊 Average Latency: {avg_latency:.2f} seconds")
print(f"💰 Total Token Consumption: {total_tokens}")
```

```python
import matplotlib.pyplot as plt
# Create a scatter plot
plt.figure(figsize=(10, 6))
plt.scatter(df['response_tokens'], df['latency'], color='blue', alpha=0.7)
# Add labels and title
plt.title('Latency vs. Response Length')
plt.xlabel('Response Tokens (Length)')
plt.ylabel('Latency (Seconds)')
plt.grid(True, linestyle='--', alpha=0.6)
# Show the plot
plt.show()
```
